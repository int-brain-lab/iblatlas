# Convert between acronym, Allen id and name

Each brain region has an acronym, an Allen id and a full name. `BrainRegions` converts between them; the output is in the order of the input, with repeats kept.

```python
from iblatlas.regions import BrainRegions

br = BrainRegions()

ids = br.acronym2id(['MOp', 'CA1', 'VISa'])
print(ids)
print(br.id2acronym(ids))
print(br.get(ids).name)
```

`br.get` returns the id, name, acronym, colour, level, parent and order of each region. Negative ids are the left hemisphere and give the same acronym; see [Tell left from right hemisphere](tell-left-from-right-hemisphere.ipynb).

## Find a region from its name

There is no lookup by full name. Search `br.name` instead, here for every region whose name contains "hippocampal", ignoring case, in the right hemisphere (positive ids) to list each region once:

```python
import numpy as np

match = (np.char.find(np.char.lower(br.name.astype(str)), 'hippocampal') >= 0) & (br.id > 0)
for acronym, name in zip(br.acronym[match], br.name[match]):
    print(acronym, name)
```

## Acronyms or ids the atlas does not have

An acronym or id that is not in the atlas is left out of the result without an error, so the output can be shorter than the input and no longer line up with it. Here `XYZ` is dropped and four ids come back for five acronyms:

```python
acronyms = ['MOp', 'CA1', 'CA1', 'XYZ', 'VISa']
ids = br.acronym2id(acronyms)
print(ids)
print(len(acronyms), len(ids))
```

Compare the lengths, or check `np.isin(acronyms, br.acronym)`, before using the result.
