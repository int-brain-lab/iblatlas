# Tell left from right hemisphere

The ML coordinate gives the hemisphere: points with a negative ML coordinate are in the left hemisphere, points with a positive one in the right. Coordinates are in metres relative to bregma, ordered ML, AP, DV; see [Coordinates and volumes](../explanation/coordinates-and-volumes.qmd).

```python
import numpy as np
import pandas as pd

from iblatlas.atlas import AllenAtlas

ba = AllenAtlas()
# Two pairs of points, each mirrored across the midline. ML, AP, DV in metres from bregma
xyz = np.array([[-0.002, -0.002, -0.0015],
                [ 0.002, -0.002, -0.0015],
                [-0.0015, -0.002, -0.0035],
                [ 0.0015, -0.002, -0.0035]])
np.where(xyz[:, 0] < 0, 'left', 'right')
```

## Lateralised mappings give left regions negative ids

Each mapping has a lateralised version whose name ends in `-lr`. It gives regions in the left hemisphere negative ids. Without `-lr`, both hemispheres share the positive id.

```python
for mapping in ('Allen', 'Allen-lr', 'Beryl-lr', 'Cosmos-lr'):
    print(f'{mapping:<9}', ba.get_labels(xyz, mapping=mapping))
```

## Acronyms drop the sign

`id2acronym` returns the same acronym for a negative id and its positive id, so the hemisphere is lost once ids become acronyms. Keep it in a column of its own:

```python
ids = ba.get_labels(xyz, mapping='Allen-lr')
pd.DataFrame({'id': ids, 'acronym': ba.regions.id2acronym(ids),
              'hemisphere': np.where(xyz[:, 0] < 0, 'left', 'right')})
```
